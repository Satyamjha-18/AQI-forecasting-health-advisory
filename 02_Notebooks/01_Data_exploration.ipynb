```python
delhi = df[df["City"] == "Delhi"].set_index("Date")

delhi["AQI"].plot(figsize=(12, 4), title="Delhi AQI (2015-2020)")
plt.ylabel("AQI")
plt.show()
```

```python
bhopal = df[df["City"] == "Bhopal"].set_index("Date")

bhopal["AQI"].plot(figsize=(12, 4), title="Bhopal AQI (2015-2020)", color="green")
plt.ylabel("AQI")
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
two_cities = df[df["City"].isin(["Delhi", "Bhopal"])].copy()
two_cities["Month"] = two_cities["Date"].dt.month

two_cities.groupby(["Month", "City"])["AQI"].mean().unstack().plot(
    kind="bar", figsize=(10, 4), title="Average AQI by Month")
plt.ylabel("Average AQI")
plt.show()
```
